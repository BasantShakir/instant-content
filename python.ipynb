```python
#Nested If statment

#inputs user
user=input("enter your username")
password=input("enter your password")

#system DB
userSystem="admin"
passwordSystem="A1234"
#هو هنا لو اليوزر نيم عندي غلط هيطبع
#Invalid username"
#طب لو صح هيخش يشوف الباسورد عندي لو صح يطبع you login successfully
#ولو غلط هيطبع"Invalid password
# يعني هو مش هيخش لل باسورد غير لما يتاكد ان username شغال
if user==userSystem:
    if password==passwordSystem:
        print("you login successfully")
    else:
        print("Invalid password")
else:
    print("Invalid username")

```

```python
#Nested If statment
#هي نفسها بس ضفت شرط في حالة لو الاتنين عندي غلط
#inputs user
user=input("enter your username")
password=input("enter your password")

#system DB
userSystem="admin"
passwordSystem="A1234"

if user==userSystem or password==passwordSystem:
  if user==userSystem:
      if password==passwordSystem:
           print("you login successfully")
      else:
        print("Invalid password")
  else:
       print("Invalid username")

else:
    print("Invaild username and password")
```

```python
#recursion ---->كأن الدالة بتنادي علي نفسها 
#هي مش loop هي function جوه function جوه function
#             5
def factorial(num):#paramter 
    if num==0 or num==1:
        return 1#استخدمنا return علشان يقدر انه يخون القيمة 
    else:
        return num*factorial(num-1)
# step1 ----> 5*fac(4)
# step2 ----> 5*4*fac(3)
# step3 ----> 5*4*3*fac(2)
# step3 ----> 5*4*3*2*fac(1)
# step3 ----> 5*4*3*2*1
#مقلش صفر لانه وقف لحد 1
```

```python
factorial(5)
```

```python
#تدخل رقم ويعملهولك بشكل تنازلي 
def count_down(n):
    if n==0:
        return 1 
    else:
        print(n)
        count_down(n-1)
count_down(10)
```

```python
#lamdaa function 
#مشكلتها انها بسيطة يعني مش بقدر اكتب فيها تفاصيل كتير بقي وتعقيدات زي ال function العادية
#ال function اللي قبل كده كانت معقدة وبتتكتب علي اكتر من سطر 
#لكن انا عاوزة اكتب ال function علي سطر واحد
#sum1 --->name function
#lambda x,y---->هتستقبل قيمتين
#: x+y --->return x+y
sum1=lambda x,y : x+y
sum1(1,2)
```
